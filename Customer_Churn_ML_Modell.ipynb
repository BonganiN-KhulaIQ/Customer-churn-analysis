## 1. Load and Explore the Dataset

The dataset used in this notebook has already been cleaned and prepared through the feature engineering process. It contains customer information together with the target variable (`churn`), which indicates whether a customer has left the company.

Before building the model, the data is loaded and inspected to ensure that it is ready for training.

```python
df = pd.read_csv('./data_for_predictions.csv')
df.drop(columns=["Unnamed: 0"], inplace=True)
df.head()
```

## 2. Prepare the Data

To train a machine learning model, the dataset is divided into two parts:

- **Training data** is used to teach the model patterns in the data.
- **Testing data** is kept separate so that the model can be evaluated on data it has not seen before.

A 75/25 split is used to provide sufficient data for training while maintaining a reliable test set for evaluation.

```python
from sklearn import metrics
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
```

## 2. Prepare the Data

To train a machine learning model, the dataset is divided into two parts:

- **Training data** is used to teach the model patterns in the data.
- **Testing data** is kept separate so that the model can be evaluated on data it has not seen before.

A 75/25 split is used to provide sufficient data for training while maintaining a reliable test set for evaluation.

```python
# Make a copy of our data
train_df = df.copy()

# Separate target variable from independent variables
y = df['churn']
X = df.drop(columns=['id', 'churn'])
print(X.shape)
print(y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
print(X_train.shape)
print(y_train.shape)
print(X_test.shape)
print(y_test.shape)
```

## 3. Train the Random Forest Model

A Random Forest Classifier is used because it combines the predictions of multiple decision trees to produce a more accurate and stable model. This algorithm is well suited to customer churn prediction as it can capture complex relationships between customer behaviour and the likelihood of churn.

```python
from sklearn.ensemble import RandomForestClassifier
# Add model training in here!
model = RandomForestClassifier(n_estimators=200, random_state=42)

model.fit(X_train, y_train)
```

## 4. Evaluate Model Performance

Once the model has been trained, its predictions are compared with the actual customer outcomes.

Several evaluation metrics are calculated to provide a balanced assessment of the model's performance. These metrics help determine not only how often the model is correct, but also how effectively it identifies customers who are likely to churn.

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    roc_auc_score
)
```

```python
y_pred = model.predict(X_test)
```

```python
print("Accuracy :", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall   :", recall_score(y_test, y_pred))
print("F1 Score :", f1_score(y_test, y_pred))
print("ROC AUC  :", roc_auc_score(y_test, model.predict_proba(X_test)[:,1]))
```

## Discussion of Evaluation Metrics

Several evaluation metrics were selected because each provides different insights into the model's performance.

- **Accuracy** measures the overall percentage of correct predictions.
- **Precision** indicates how many customers predicted to churn actually churned.
- **Recall** measures how effectively the model identifies customers who genuinely churn.
- **F1-score** provides a balanced measure of Precision and Recall, making it useful when both false positives and false negatives are important.
- **ROC-AUC** evaluates the model's ability to distinguish between customers who churn and those who remain.
- **Confusion Matrix** provides a detailed summary of correct and incorrect classifications.

Using multiple metrics provides a more complete understanding of the model than relying on accuracy alone.

```python
print(confusion_matrix(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

The Random Forest Classifier was successfully trained to predict customer churn using the prepared dataset. The evaluation metrics provide an indication of how well the model performs on unseen data and highlight its strengths in identifying customers who are likely to churn.
